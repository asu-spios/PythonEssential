## 7. Розбір рядка: strptime та помилки

strptime виконує зворотне перетворення: читає рядок за заданим шаблоном і створює datetime. Послідовність роздільників і порядок полів мусять відповідати формату; для невірної дати або невідповідного рядка виникає ValueError. Якщо вхід походить від користувача, обробляйте цю помилку, а не припускайте, що кожен текст є правильною датою.

```python
from datetime import datetime

for raw in ["14.03.2026", "31.02.2026"]:
    try:
        parsed = datetime.strptime(raw, "%d.%m.%Y")
    except ValueError:
        print("Некоректна дата:", raw)
    else:
        print(parsed.date())
```

## 8. ISO 8601 та fromisoformat

ISO подання полегшує обмін даними: дата записується як YYYY-MM-DD, час — після T, а зсув зони за потреби додається в кінці. isoformat() створює рядок, fromisoformat() читає сумісний рядок. Наївний час без зсуву не повідомляє, в якій зоні його слід тлумачити. Відтворюємо саме той об’єкт, який створили, і порівнюємо їх.

```python
from datetime import datetime, timezone

moment = datetime(2026, 4, 2, 16, 45, tzinfo=timezone.utc)
encoded = moment.isoformat()
decoded = datetime.fromisoformat(encoded)
print(encoded)                    # 2026-04-02T16:45:00+00:00
print(decoded == moment)          # True
```

Для перетворення використовуйте datetime з явно заданою зоною: timestamp() дістає числову мітку, fromtimestamp(value, tz=timezone.utc) повертає час у UTC.
```python
from datetime import datetime, timezone

moment = datetime(2026, 1, 1, 0, 0, tzinfo=timezone.utc)
stamp = moment.timestamp()
restored = datetime.fromtimestamp(stamp, tz=timezone.utc)
print("Timestamp:", stamp)
print("UTC:", restored, "Рівні:", restored == moment)
```

```python
from datetime import datetime, timezone
from zoneinfo import ZoneInfo

kyiv = ZoneInfo("Europe/Kyiv")
winter = datetime(2026, 1, 15, 12, 0, tzinfo=timezone.utc)
summer = datetime(2026, 7, 15, 12, 0, tzinfo=timezone.utc)
print("Зима:", winter.astimezone(kyiv).isoformat())
print("Літо:", summer.astimezone(kyiv).isoformat())
print("Початковий момент UTC:", winter.isoformat())
```

## 11. Модуль time: поточна мітка, localtime і gmtime

time.time() повертає поточну числову мітку; значення щомиті змінюється. time.localtime(t) подає її як структуру з полями дати і часу в зоні середовища, time.gmtime(t) — у UTC. time.ctime(t) створює читабельний локальний рядок; формат і назви можуть залежати від системи. Це різні способи подати той самий момент, а не різні моменти.

```python
import time

stamp = time.time()
local = time.localtime(stamp)
utc = time.gmtime(stamp)
print("Мітка:", stamp)
print("Рік UTC:", utc.tm_year, "день року:", utc.tm_yday)
print("Локальна година:", local.tm_hour)
print("Рядок:", time.ctime(stamp))
```

## 14. choice і shuffle

choice(seq) повертає один елемент непорожньої послідовності. shuffle(list) змінює порядок елементів самого списку й повертає None, а не новий список. Якщо первісний порядок потрібен надалі, спочатку створіть копію. Окремий генератор допомагає повторювати експеримент, але порядок залежить від того, скільки випадкових викликів уже було зроблено.

```python
import random

rng = random.Random(12)
cards = ["Туз", "Король", "Дама", "Валет"]
deck = cards.copy()
picked = rng.choice(deck)
rng.shuffle(deck)
print("Вибрано:", picked)
print("Перемішано:", deck)
print("Оригінал:", cards)
assert sorted(deck) == sorted(cards)
```

## 15. sample та choices: без повторів і з повторами

sample(population, k) обирає k різних елементів без повернення; k не може перевищити розмір популяції. choices(population, k=k) робить вибір із поверненням, тому результати можуть повторюватися. Параметр weights надає відносні ваги; наприклад [8, 1, 1] означає, що перший варіант має 8 із 10 часток загальної ваги, а не гарантію конкретного результату в десяти спробах.

```python
import random

rng = random.Random(123)
participants = ["А", "Б", "В", "Г", "Д"]
team = rng.sample(participants, k=3)
draws = rng.choices(["червоний", "синій"], weights=[3, 1], k=8)
print("Команда:", team)
print("Витягування:", draws)
assert len(team) == len(set(team)) == 3
```
