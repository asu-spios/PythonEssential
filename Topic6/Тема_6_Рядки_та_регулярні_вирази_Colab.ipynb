```python
record = "WARN|disk|WARN"
print(record.find("WARN"))   # 0
print(record.rfind("WARN"))  # 10
print(record.find("ERROR"))  # -1
try:
    print(record.index("ERROR"))
except ValueError:
    print("Обов'язкового маркера немає")
```

## 7. Параметри URL: спеціалізований розбір

Рядок запиту URL має власні правила кодування: пробіл може бути записаний як + або %20, а символи даних — у відсотковому кодуванні. Ручний split за & та = легко помилиться, якщо значення містить закодовані символи або повторний ключ. Стандартний urllib.parse розбирає ці правила. parse_qs повертає список значень для кожного ключа, бо ключ може повторюватися. Це продовжує приклад документа, але показує безпечніший інструмент для реального URL.

```python
from urllib.parse import urlsplit, parse_qs

url = "https://example.org/search?q=мережна+безпека&tag=python&tag=regex"
query = urlsplit(url).query
params = parse_qs(query)
print(params)
print(params["q"][0])   # мережна безпека
print(params["tag"])    # два значення
```

## 13. Сирі рядки та метасимволи re

Модуль re читає шаблон як окрему мову. У Python літерал r"\d+" зберігає зворотну риску без додаткового екранування, тому для regex зазвичай беруть сирі рядки. \d означає цифру, крапка . — один довільний символ (крім нового рядка за замовчуванням), а \. — саме крапку. Якщо потрібен буквальний фрагмент, сформований із даних користувача, re.escape екранує його спеціальні символи.

```python
import re

text = "Версії: app.1 та appX1"
print(re.findall(r"app.1", text))       # крапка як довільний символ
print(re.findall(r"app\.1", text))     # лише буквальна крапка
literal = "app.1"
print(re.findall(re.escape(literal), text))
```

## 14. Класи символів і Unicode

Клас [A-Z] означає одну латинську велику літеру, [0-9] — одну цифру, а [^:] — будь-який символ, крім двокрапки. Скорочення \w у Python за замовчуванням охоплює Unicode-літери й цифри, а також підкреслення; воно не обмежується латиницею. Так само \d може збігатися з Unicode-десятковими цифрами. Якщо формат вимагає саме ASCII, пишіть [0-9] або використовуйте прапорець re.ASCII. Вибір класу залежить від вимоги до даних.

```python
import re

text = "UA-42; DE-7; Україна-9"
print(re.findall(r"[A-Z]{2}-[0-9]+", text))  # ['UA-42', 'DE-7']
print(re.findall(r"\w+", text))              # і кириличне слово
print(re.findall(r"[^;]+", text))             # сегменти між ;
```

## 18. search, Match.group і span

re.search повертає перший об’єкт Match або None. group() дає знайдений текст, span() — пару індексів початку та кінця; кінцевий індекс не входить у зріз. Перед викликом group перевіряйте, що match існує. Це особливо корисно для пошуку часу чи коду у вільному тексті. Пошук не підтверджує коректність усього повідомлення: для повного формату потрібен fullmatch.

```python
import re

message = "gateway-07 відповів за 125 ms"
match = re.search(r"\b[0-9]+\s*ms\b", message)
if match:
    print("Знайдено:", match.group())
    print("Позиції:", match.span())
    print("Зріз:", message[match.start():match.end()])
```

## 19. findall і finditer

findall повертає список усіх неперекривних збігів. Якщо в шаблоні є захоплювальні групи, елементами можуть бути групи або кортежі груп замість повного тексту. finditer повертає об’єкти Match один за одним, тому підходить, коли потрібні і групи, і позиції. Для повторюваних токенів це допомагає показати, де саме кожен трапився.

```python
import re

text = "CPU=81; RAM=64; CPU=77"
pattern = r"([A-Z]+)=([0-9]+)"
print("Групи findall:", re.findall(pattern, text))
for match in re.finditer(pattern, text):
    print(match.group(0), match.groups(), match.span())
```

## 23. Прапорці IGNORECASE та MULTILINE

re.IGNORECASE дозволяє пошук без урахування регістру. re.MULTILINE змінює значення якорів ^ та $ так, щоб вони працювали на початку й у кінці кожного рядка, а не тільки всього тексту. Це не змінює поведінки крапки щодо перенесення рядка: для цього існує DOTALL. Прапорці можна об’єднувати оператором |. На прикладі журналу знайдемо рядки, що починаються з ERROR у різних регістрах.

```python
import re

log = "INFO ready\nerror disk\nWARN slow\nERROR network"
matches = re.findall(r"^error\s+.+$", log,
                     flags=re.IGNORECASE | re.MULTILINE)
print(matches)  # два рядки з помилками
```
