## 5. Незмінні й змінні об’єкти як аргументи

Під час виклику ім’я параметра посилається на той самий об’єкт, що й переданий аргумент. Рядок не можна змінити на місці: присвоєння нового рядка локальному імені не змінює зовнішню змінну. Список змінюваний, тому append змінює об’єкт, який бачить і викликач. Якщо треба зберегти початковий список, створіть його копію й поверніть новий результат. list.copy() є поверхневою копією: вкладені змінювані об’єкти залишаться спільними.

```python
def renamed(text):
    text = "новий"
    return text

def add_item_in_place(items, item):
    items.append(item)

def with_item(items, item):
    result = items.copy()
    result.append(item)
    return result

label = "старий"
numbers = [1, 2]
print(renamed(label), label)
add_item_in_place(numbers, 3)
print("Після append:", numbers)
new_numbers = with_item(numbers, 4)
print("Оригінал:", numbers, "Копія:", new_numbers)
```

## 6. Функція, що поєднує цикл і словник

ord(char) повертає кодову позицію символу в Unicode. Для латинської літери A це 65, але для українських літер значення не є кодом ASCII, бо ASCII не містить кирилиці. Словник зіставляє кожен унікальний символ із його числовим кодом. Якщо символ повторюється, він уже є ключем і додавати його знову не потрібно.

```python
def string_to_codes(text: str) -> dict[str, int]:
    codes = {}
    for char in text:
        if char not in codes:
            codes[char] = ord(char)
    return codes

print(string_to_codes("Hello"))
print("Код літери А:", ord("А"))  # кирилична А
```

## 8. Вкладена функція та nonlocal

Вкладена функція бачить імена зі своєї охоплювальної функції. Щоб переприсвоїти ім’я count з цієї зовнішньої функції, усередині вкладеної функції пишуть nonlocal count. Без nonlocal присвоєння створило б нове локальне ім’я. Замикання зберігає доступ до count між викликами поверненої функції. Кожен окремий виклик фабрики лічильників створює власний стан.

```python
def make_counter():
    count = 0
    def next_count():
        nonlocal count
        count += 1
        return count
    return next_count

counter = make_counter()
print(counter(), counter(), counter())  # 1 2 3
other_counter = make_counter()
print(other_counter())  # 1
```

Для незалежного списку на кожний виклик використайте None як сигнал і створюйте список усередині функції. Якщо передано готовий список, цей приклад його змінює — це окрема свідома поведінка.

```python
def collect(item, basket=None):
    if basket is None:
        basket = []
    basket.append(item)
    return basket

print(collect("яблуко"))  # ['яблуко']
print(collect("груша"))   # ['груша']
shared = ["хліб"]
print(collect("молоко", shared))
print("Переданий список:", shared)
```

Параметр з однією зірочкою збирає додаткові позиційні аргументи в кортеж. Ім’я args є домовленістю, а спеціальне значення має саме зірочка.
```python
def add_all(*numbers: float) -> float:
    total = 0
    for number in numbers:
        total += number
    return total

print(add_all())         # 0
print(add_all(2, 3, 5))  # 10
```

## 13. Іменовані аргументи **kwargs та поєднання

Параметр із двома зірочками збирає додаткові іменовані аргументи у словник. У визначенні функції *args ставлять перед **kwargs, щоб правильно розподілити аргументи. Згодом можна пройтися за items() і обробити кожну пару. Коли структура даних наперед відома, звичайні явні параметри часто простіше читати; kwargs корисний для гнучкого набору полів.

```python
def describe_event(*participants, **details):
    print("Учасники:", participants)
    for key, value in details.items():
        print(f"{key}: {value}")

describe_event("Оля", "Тарас", room="204", duration=45)
```

## 15. Рекурсія і базовий випадок

Рекурсивна функція викликає себе для меншої підзадачі. Потрібні базовий випадок, який завершує виклики, і крок, який до нього наближає. Факторіал 0 дорівнює 1, а для n > 0 маємо n! = n × (n-1)!. Для від’ємних значень цей приклад явно викликає ValueError, щоб не заглиблюватися без кінця. Кожен виклик має власні параметри в стеку викликів.

```python
def factorial(n: int) -> int:
    if n < 0:
        raise ValueError("n має бути невід’ємним")
    if n == 0:
        return 1
    return n * factorial(n - 1)

print("0! =", factorial(0))
print("5! =", factorial(5))  # 120
```

## 16. Стек викликів та числа Фібоначчі

Коли функція викликає себе, поточний виклик чекає на результат наступного. Досягнувши базового випадку, виклики повертаються у зворотному порядку. Пряма рекурсивна формула Фібоначчі F(n)=F(n-1)+F(n-2) добре показує цю ідею, але повторно обчислює ті самі значення та швидко сповільнюється. Для великих n доречні цикл або кешування; тут беремо мале n. Ліміт глибини рекурсії залежить від середовища, тому не слід покладатися на фіксоване число.

```python
def fibonacci(n: int) -> int:
    if n < 0:
        raise ValueError("n має бути невід’ємним")
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print([fibonacci(n) for n in range(8)])
# [0, 1, 1, 2, 3, 5, 8, 13]
```
