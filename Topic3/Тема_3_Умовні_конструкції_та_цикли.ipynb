## 1. Умовне виконання: `if`, `else` та порівняння

Умова має давати логічний результат `True` або `False`. Оператор `if` виконує свій блок, коли умова істинна; `else` задає альтернативну гілку.

Оператори порівняння: `==` дорівнює, `!=` не дорівнює, `<`, `<=`, `>`, `>=`. Не плутайте `=` (присвоєння) з `==` (порівняння).


```python
temperature = 18

if temperature >= 20:
    print("Тепло: можна гуляти без куртки.")
else:
    print("Прохолодно: візьміть куртку.")
```

**Пояснення.** Python перевіряє `temperature >= 20`. Оскільки для значення 18 результатом є `False`, виконується гілка `else`. Змініть значення змінної та запустіть комірку повторно.

## 2. Кілька варіантів: `elif`

`elif` (скорочення від *else if*) додає наступну умову. Умови перевіряються згори вниз; виконується лише перша істинна гілка. Тому спочатку варто перевіряти конкретніші випадки.


```python
score = 84

if score >= 90:
    grade = "відмінно"
elif score >= 75:
    grade = "добре"
elif score >= 60:
    grade = "задовільно"
else:
    grade = "потрібно повторити матеріал"

print(f"Оцінка: {grade}")
```

**Пояснення.** Значення 84 не проходить першу умову, але проходить `score >= 75`, тому нижчі гілки вже не перевіряються. Якщо поставити загальну умову `score > 0` перед перевіркою `score == 100`, випадок 100 буде перехоплено раніше.

## 3. Логічні оператори та істинність значень

- `and` — істина, якщо істинні обидві умови;
- `or` — істина, якщо істинна хоча б одна умова;
- `not` — заперечує логічне значення.

У Python нуль, `None`, порожній рядок і порожні колекції (`[]`, `{}`) у логічному контексті вважаються хибними. Непорожні значення зазвичай істинні.


```python
age = 19
has_ticket = True

if age >= 18 and has_ticket:
    print("Доступ дозволено")
else:
    print("Доступ заборонено")

name = ""
if name:
    print(f"Вітаємо, {name}!")
else:
    print("Ім’я не введено")
```

**Пояснення.** Для доступу мають одночасно виконуватися обидві умови. Перевірка `if name:` зручна для з’ясування, чи рядок не порожній.

### `==` і `is` — різні перевірки

`==` порівнює значення об’єктів. `is` перевіряє, чи це той самий об’єкт. Для перевірки відсутності значення використовуйте `is None`.

```python
result = None

if result is None:
    print("Результату ще немає")
```

## 4. Цикл `for` і функція `range()`

Цикл `for` проходить по елементах послідовності. `range(stop)` створює числа від 0 до `stop - 1`; `range(start, stop, step)` задає початок, межу (не включається) та крок.


```python
for number in range(1, 6):
    print(number, "у квадраті дорівнює", number ** 2)
```

**Пояснення.** Цикл виконується п’ять разів для значень 1, 2, 3, 4, 5. Межа 6 не включається. Зверніть увагу на відступ: він визначає тіло циклу.

### Перебір списку, `enumerate()` і `zip()`

```python
tasks = ["прочитати тему", "виконати приклад", "перевірити результат"]

for index, task in enumerate(tasks, start=1):
    print(f"{index}. {task}")
```

`enumerate()` дає індекс і значення. `zip()` об’єднує відповідні елементи кількох послідовностей і зупиняється на найкоротшій.

```python
students = ["Олена", "Ігор", "Марія"]
scores = [92, 78, 88]

for student, score in zip(students, scores):
    print(f"{student}: {score} балів")
```

## 5. Цикл `while`

`while` повторює блок, доки умова істинна. У тілі циклу потрібно змінювати стан, від якого залежить умова, інакше цикл може стати нескінченним.


```python
countdown = 3
while countdown > 0:
    print(countdown)
    countdown -= 1
print("Старт!")
```

**Пояснення.** На кожній ітерації значення `countdown` зменшується на 1. Коли воно стає 0, умова хибна і цикл завершується.

## 6. Керування циклом: `break` і `continue`

`break` негайно завершує цикл. `continue` пропускає решту поточної ітерації та переходить до наступної.

```python
# Знайти перше число, що ділиться на 7
for number in range(1, 30):
    if number % 7 == 0:
        print("Перше кратне 7:", number)
        break
```

```python
# Вивести лише непарні числа
for number in range(1, 8):
    if number % 2 == 0:
        continue
    print(number)
```

## 7. Цикли та словники

За словником можна перебирати ключі, значення або пари ключ–значення за допомогою `.keys()`, `.values()` і `.items()`. Не додавайте й не видаляйте елементи колекції під час її прямого перебору.


```python
inventory = {"зошит": 12, "ручка": 8, "олівець": 5}

for item, quantity in inventory.items():
    print(f"{item}: {quantity} шт.")
```

## 8. Обробка помилок: `try...except`

Перетворення введеного рядка на число може завершитися помилкою `ValueError`. Блок `try` містить потенційно проблемну операцію, а `except` описує реакцію на конкретну помилку.

У Colab функція `input()` працює у комірці, однак для повторюваних демонстрацій зручніше змінювати заданий рядок у прикладі.


```python
raw_value = "25"  # спробуйте замінити на "abc"

try:
    age = int(raw_value)
except ValueError:
    print(f"Значення {raw_value!r} не є цілим числом")
else:
    if age >= 18:
        print("Повнолітній вік")
    else:
        print("Неповнолітній вік")
finally:
    print("Перевірку завершено")
```

**Пояснення.** Якщо `int(raw_value)` викликає `ValueError`, виконується `except`. Блок `else` виконується лише за відсутності помилки, а `finally` — у будь-якому разі.

## 9. Розгорнутий приклад: підрахунок успішних оцінок

У прикладі програма проходить список оцінок, пропускає некоректні значення, рахує кількість результатів від 60 балів і визначає середнє значення коректних оцінок.

```python
raw_scores = ["95", "72", "помилка", "58", "86"]
valid_scores = []
passed_count = 0

for raw_score in raw_scores:
    try:
        score = int(raw_score)
    except ValueError:
        print(f"Пропущено некоректне значення: {raw_score!r}")
        continue

    if not 0 <= score <= 100:
        print(f"Оцінка поза межами 0–100: {score}")
        continue

    valid_scores.append(score)
    if score >= 60:
        passed_count += 1

if valid_scores:
    average = sum(valid_scores) / len(valid_scores)
    print(f"Коректних оцінок: {len(valid_scores)}")
    print(f"Середній бал: {average:.1f}")
    print(f"Успішних оцінок: {passed_count}")
else:
    print("Немає коректних оцінок для аналізу")
```

**Як працює приклад:** `for` перебирає дані; `try...except` відкидає значення, яке не перетворюється на число; `continue` переходить до наступного елемента; `if` перевіряє межі й рахує успішні результати. Перевірте роботу, змінивши список `raw_scores`.
