## 6. match і case

match зіставляє значення з шаблонами case, а case _ є запасною гілкою. Для кількох фіксованих команд це зручно замість довгого ланцюга elif. Конструкція потребує Python 3.10 або новішого.

```python
command = "start"
match command:
    case "start":
        print("Запуск")
    case "stop":
        print("Зупинка")
    case _:
        print("Невідома команда")
```

## 9. break і continue

continue переходить до наступної ітерації, а break завершує увесь цикл. Якщо використовувати continue у while, оновіть лічильник до переходу; інакше умова може залишитися незмінною.

```python
values = [5, 0, 8, -1, 12]
for value in values:
    if value < 0:
        break
    if value == 0:
        continue
    print("Прийнято:", value)
# 5, 8
```

## 13. Обхід словника

Звичайний цикл за словником або .keys() дає ключі. .values() дає значення, а .items() — пари ключів і значень. Під час прямого обходу не додавайте і не видаляйте ключі, хоча присвоєння нового значення за вже наявним ключем можливе.

```python
inventory = {"кабель": 5, "датчик": 2, "шлюз": 1}
for name, amount in inventory.items():
    print(name, amount)
for name in inventory:
    inventory[name] += 1
print(inventory)
```

## 14. try та except

int для некоректного тексту викликає ValueError; ділення на нуль — ZeroDivisionError. Розмістіть ризиковану операцію у try та обробляйте конкретну помилку в except. Синтаксичну помилку слід виправляти в коді, а не маскувати обробником.

```python
for raw in ["12", "помилка", "0"]:
    try:
        number = int(raw)
        result = 24 / number
    except ValueError:
        print(raw, "— не число")
    except ZeroDivisionError:
        print(raw, "— ділення на нуль")
    else:
        print("Результат:", result)
```

## 15. else і finally

else після try/except працює лише за відсутності винятку. finally виконується після кожної спроби, тому підходить для завершальних дій. Залишайте в try лише те, що може спричинити саме очікуваний виняток.

```python
for raw in ["5", "невідомо"]:
    try:
        quantity = int(raw)
    except ValueError:
        print("Помилка:", raw)
    else:
        print("Прийнято:", quantity)
    finally:
        print("Спробу завершено")
```
