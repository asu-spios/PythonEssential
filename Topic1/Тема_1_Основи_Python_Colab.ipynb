## 6. Вирази й арифметичні оператори

Вираз обчислює значення. Доступні `+`, `-`, `*`, `/` (звичайне ділення), `//` (ділення з округленням униз), `%` (остача), `**` (степінь). `//` не просто відкидає дробову частину: наприклад, `-7 // 2 == -4`. Ділення `/` повертає `float`.

```python
a, b = 7, 2
print("Сума:", a + b, "Добуток:", a * b)
print("Ділення:", a / b, "Ціла частка:", a // b, "Остача:", a % b)
print("Степінь:", a ** b)
print("Для від’ємного:", -7 // 2)
```

### Приклад: години, хвилини, секунди
Залишок `%` допомагає розкласти 5000 секунд на складники. Перевірка наприкінці відновлює початкове значення.

```python
total_seconds = 5000
hours = total_seconds // 3600
minutes = (total_seconds % 3600) // 60
seconds = total_seconds % 60
print(f"{hours} год {minutes} хв {seconds} с")
assert hours * 3600 + minutes * 60 + seconds == total_seconds
```

## 7. Порядок операцій та геометричні задачі

Спочатку обчислюються дужки, далі піднесення до степеня, потім множення та ділення, після них додавання й віднімання. Для від’ємного числа `-3 ** 2` дає `-9`, а `(-3) ** 2` дає `9`. Явні дужки роблять формули зрозумілішими.

```python
print(-3 ** 2, (-3) ** 2)
side_a, side_b = 10, 5
hypotenuse = (side_a ** 2 + side_b ** 2) ** 0.5
area = side_a * side_b / 2
print(f"Гіпотенуза: {hypotenuse:.3f}; площа: {area:.1f}")
```

### Приклад: відстань між точками
Формула: $d = \sqrt{(x_2-x_1)^2 + (y_2-y_1)^2}$. Формат `{distance:.2f}` змінює **показ** результату, а не збережене число.

```python
x1, y1 = 10, 10
x2, y2 = 25, 25
distance = ((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5
print(f"Відстань: {distance:.2f}")
```
